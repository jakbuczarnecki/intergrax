# dispute_intake experiment

Run via NexusLoop or ``applications/lab_application``.

```python
from dispute_intake.dispute_intake_agent import DisputeIntakeAgent
from intergrax.runtime.nexus.nexus_loop import NexusLoop
from intergrax.runtime.registry.agent_registry import AgentRegistry
from intergrax.runtime.task.task import Task, TaskContext

registry = AgentRegistry()
registry.register(DisputeIntakeAgent())
loop = NexusLoop(registry)
task = Task(tenant_id='t1', user_id='u1', message='hello', context=TaskContext(capability='dispute.intake'))
result = await loop.handle_task(task)
result.answer
```

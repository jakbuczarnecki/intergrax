### Fill VectorStore

```python
from openai import OpenAI
from dotenv import load_dotenv
import os

from intergrax.tools.providers.openai_vector_store.service import OpenAIVectorStoreOps

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
VECTOR_STORE_ID = os.getenv("VECTOR_STORE_ID")

# use your folder with data (pdf, txt, docx)
LOCAL_FOLDER = "../temp_documents/mooff-strategy/txt"

client = OpenAI()
ops = OpenAIVectorStoreOps(client=client, vector_store_id=VECTOR_STORE_ID)
ops.ensure_vector_store_exists()
ops.clear_vector_store_and_storage()
uploaded, failed = ops.upload_folder(LOCAL_FOLDER)
print(f"uploaded={len(uploaded)} failed={len(failed)}")
```
